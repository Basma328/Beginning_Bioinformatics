### Integer (`int`)
Integers are whole numbers, positive or negative, without decimals.

````python
my_integer = 42
print(f"Value: {my_integer}, Type: {type(my_integer)}")
````

### Floating-Point Number (`float`)
Floats are numbers with a decimal point.

````python
my_float = 3.14159
print(f"Value: {my_float}, Type: {type(my_float)}")
````

### String (`str`)
Strings are sequences of characters, enclosed in single or double quotes.

````python
my_string_example = "Hello Python!"
print(f"Value: '{my_string_example}', Type: {type(my_string_example)}")
````

### Boolean (`bool`)
Booleans represent truth values: `True` or `False`.

````python
my_boolean = True
print(f"Value: {my_boolean}, Type: {type(my_boolean)}")
````

### List (`list`)
Lists are ordered, mutable (changeable) collections of items, enclosed in square brackets.

````python
my_list_example = [1, "apple", True, 3.5]
print(f"Value: {my_list_example}, Type: {type(my_list_example)}")
````

### Tuple (`tuple`)
Tuples are ordered, immutable (unchangeable) collections of items, enclosed in parentheses.

````python
my_tuple = (10, "banana", False)
print(f"Value: {my_tuple}, Type: {type(my_tuple)}")
````

### Dictionary (`dict`)
Dictionaries are unordered collections of key-value pairs, enclosed in curly braces.

````python
my_dictionary = {"name": "Alice", "age": 30}
print(f"Value: {my_dictionary}, Type: {type(my_dictionary)}")
````

### Set (`set`)
Sets are unordered collections of unique items, also enclosed in curly braces (but without key-value pairs).

````python
my_set = {1, 2, 3, 2, 1}
print(f"Value: {my_set}, Type: {type(my_set)}")
````

####**TASK P19.2** Modify your code for the Rosalind #5 Dictionary problem so that it uses count( ) to count the words.

````python
# Open the file
file = open("rosalind_ini6.txt")

# Read the text and split it into words
text = file.read()
words = text.split()

# Create a dictionary
counts = {}

# Count each word using count()
for word in words:
    counts[word] = words.count(word)

# Print each word and its count
for word in counts:
    print(word, counts[word])

# Close the file
file.close()
````
